Q3. Replace missing values in the Quantity and Price columns with their respective mean values using Pandas.

```python
1. Replace Missing Values with Mean Values Using Pandas

. Use the following Python code to replace missing values in the Quantity and Price columns with their respective mean values.


import pandas as pd

# Load the dataset
df = pd.read_csv("retail_dataset (2).csv")

# Replace missing Quantity values with the mean
df["Quantity"] = df["Quantity"].fillna(df["Quantity"].mean())

# Replace missing Price values with the mean
df["Price"] = df["Price"].fillna(df["Price"].mean())

# Display the updated columns
print(df[["Quantity", "Price"]])

# Check missing values after replacement
print("\nMissing Values After Replacement:")
print(df[["Quantity", "Price"]].isnull().sum())

2. Explanation

- df["Quantity"].mean() calculates the average quantity.
- df["Price"].mean() calculates the average price.
- fillna() replaces missing values with the calculated mean.
- isnull().sum() checks whether any missing values remain in these columns.

. Conclusion: Missing values in the Quantity and Price columns are replaced with their respective mean values, helping prepare the dataset for further analysis.
```

```python
1. Remove Rows with Missing Product Category or Region

. Use the following Python code to remove all rows where either the Product Category or Region column contains missing values.


import pandas as pd

# Load the dataset
df = pd.read_csv("retail_dataset (3).csv")

# Display rows before removal
print("Rows before removal:", len(df))

# Remove rows with missing Product Category or Region
df = df.dropna(subset=["Product Category", "Region"])

# Display rows after removal
print("Rows after removal:", len(df))

# Check missing values in the selected columns
print("\nMissing Values:")
print(df[["Product Category", "Region"]].isnull().sum())


2. Explanation

- dropna(subset=["Product Category", "Region"]) removes any row with a missing value in either of these two columns.
- len(df) displays the number of rows before and after removal.
- isnull().sum() verifies that no missing values remain in these columns.

. Conclusion: All rows with missing values in Product Category or Region are removed, leaving only rows with valid values in both columns.
```

Q8. Based on the revenue calculated for each product category, identify the top 3 categories with the highest revenue and the bottom 3 categories with the lowest revenue.

```python
1. Identify Top 3 and Bottom 3 Product Categories by Revenue

. Use Pandas groupby(), sort_values(), head(), and tail() to find the three highest- and lowest-revenue categories.


import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv("retail_dataset (7).csv")

# Replace missing values with mean
df["Quantity"] = df["Quantity"].fillna(df["Quantity"].mean())
df["Price"] = df["Price"].fillna(df["Price"].mean())

# Remove rows with missing Product Category or Region
df = df.dropna(subset=["Product Category", "Region"])

# Calculate Revenue
df["Revenue"] = np.multiply(df["Quantity"], df["Price"])

# Calculate total revenue by category
category_revenue = df.groupby("Product Category")["Revenue"].sum()

# Top 3 categories
top_3 = category_revenue.sort_values(ascending=False).head(3)

# Bottom 3 categories
bottom_3 = category_revenue.sort_values(ascending=True).head(3)

print("Top 3 Categories by Revenue:")
print(top_3)

print("\nBottom 3 Categories by Revenue:")
print(bottom_3)

Output
Top 3 categories with the highest revenue
Product Category	Total Revenue
Furniture	₹129,664.53
Electronics	₹93,204.32
Clothing

Bottom 3 categories with the lowest revenue
Product Category	Total Revenue
Groceries	₹11,978.00
Beauty	₹27,647.00
Books

2. Conclusion

- Highest revenue: Furniture, generating approximately ₹129,664.53.
- Lowest revenue: Groceries, generating ₹11,978.00.
- The business could investigate why Furniture performs strongly and explore ways to improve Grocery sales.
```

Q9. Group the dataset by Region and calculate the total revenue generated in each region. Identify the region with the highest revenue and the region with the lowest revenue.

```python
1. Calculate Total Revenue by Region

. Use Pandas groupby() and NumPy to calculate the total revenue for each region and identify the highest- and lowest-revenue regions.


import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv("retail_dataset (8).csv")

# Replace missing values with mean
df["Quantity"] = df["Quantity"].fillna(df["Quantity"].mean())
df["Price"] = df["Price"].fillna(df["Price"].mean())

# Remove rows with missing Product Category or Region
df = df.dropna(subset=["Product Category", "Region"])

# Calculate Revenue
df["Revenue"] = np.multiply(df["Quantity"], df["Price"])

# Calculate total revenue by region
region_revenue = df.groupby("Region")["Revenue"].sum()

# Display revenue by region
print("Total Revenue by Region:")
print(region_revenue.sort_values(ascending=False))

# Identify highest and lowest revenue regions
print("\nRegion with Highest Revenue:")
print(region_revenue.idxmax(), region_revenue.max())

print("\nRegion with Lowest Revenue:")
print(region_revenue.idxmin(), region_revenue.min())


Output
Region	Total Revenue
North	₹153,244.35
East	₹126,232.04
South	₹104,855.74
West	₹81,197.53

2. Conclusion

- Highest revenue region: North — ₹153,244.35
- Lowest revenue region: West — ₹81,197.53

. The North region generates the most revenue, while the West region generates the least. The business could investigate sales opportunities in the West region to improve its performance.
```

Q10. Extract the month from the Date column using Pandas and group the dataset by month to calculate the total monthly revenue. Also, calculate the mean, median, and standard deviation of the Revenue column
using NumPy.

```python
1. Monthly Revenue and Statistical Analysis Using Pandas and NumPy

. Use Pandas to extract the month from the Date column and calculate total revenue for each month. Then use NumPy to calculate the mean, median, and standard deviation of the Revenue column.


import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv("retail_dataset (9).csv")

# Handle missing values
df["Quantity"] = df["Quantity"].fillna(df["Quantity"].mean())
df["Price"] = df["Price"].fillna(df["Price"].mean())

# Remove rows with missing Product Category or Region
df = df.dropna(subset=["Product Category", "Region"]).copy()

# Convert Date column into datetime format
df["Date"] = pd.to_datetime(df["Date"], format="%d-%m-%Y")

# Create Revenue column
df["Revenue"] = np.multiply(df["Quantity"], df["Price"])

# Extract month from Date
df["Month"] = df["Date"].dt.month_name()

# Calculate total monthly revenue in calendar order
monthly_revenue = df.groupby(df["Date"].dt.month)["Revenue"].sum()
monthly_revenue.index = pd.to_datetime(
    monthly_revenue.index, format="%m"
).strftime("%B")

print("Total Monthly Revenue:")
print(monthly_revenue.round(2))

# Calculate statistical measures using NumPy
revenue = df["Revenue"].to_numpy()

print("\nMean Revenue:", round(np.mean(revenue), 2))
print("Median Revenue:", round(np.median(revenue), 2))
print("Standard Deviation:", round(np.std(revenue), 2))


Output
1. Total Monthly Revenue
Month	Total Revenue
January	43,994.00
February	33,028.00
March	10,557.00
April	57,298.53
May	27,915.00
June	28,190.00
July	47,851.29
August	76,958.00
September	30,719.11
October	38,017.00
November	27,662.74
December	43,339.00

2. Statistical Measures of Revenue
Measure	Value
Mean Revenue	4,900.31
Median Revenue	3,282.11
Standard Deviation	4,721.17

2. Explanation

- dt.month_name() extracts the month name from the date.
- groupby() groups transactions by month and .sum() calculates monthly revenue.
- np.mean() calculates the average revenue.
- np.median() calculates the middle revenue value.
- np.std() calculates the population standard deviation.

. Conclusion: August generated the highest monthly revenue (76,958.00), while March generated the lowest (10,557.00). The mean revenue is higher than the median, suggesting that some transactions have relatively high revenue values.
```
